# 3 · Promotion effectiveness
- **Baseline** (BR-03): average units of clean non-promo weeks within ±8 weeks.
- **Uplift** = units / baseline - 1.
- **ROI** = (profit with promo - profit without) / discount cost, at an **ASSUMED** 30% margin (BR-09).

```python
import os, sys
from pathlib import Path
# run from the repository root so relative paths work
if Path.cwd().name == "notebooks":
    os.chdir("..")
sys.path.insert(0, "src")
import pandas as pd, numpy as np
pd.set_option("display.width", 160, "display.max_columns", 20)
```

## Worked example (acceptance criterion AC-03.2)

```python
m, base, price, units, baseline = 0.30, 4.00, 3.00, 100, 40
c = base * (1 - m)
net = units * (price - c) - baseline * (base - c)
cost = (base - price) * units
print(f'net incremental profit = {net:.2f}, discount cost = {cost:.2f}, ROI = {net / cost:.0%}')
```
